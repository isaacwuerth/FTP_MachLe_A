## Exercise 1: Data Loading, Inspection and Visualization
For this exercise, we use NBA data `nba_stats.csv` provided.


```python
# Import requiered packages
import pandas as pd
import matplotlib.pyplot as plt
```

#### 1. Load the dataset.

```python
df = pd.read_csv("nba_stats.csv", low_memory=False)
```

#### 2. Get an overview of the dataset

```python
df.info()
```

```python
df.describe()
```

###### 1. Find the average age of the players for the year 2017.

```python
df.loc[df["Year"] == 2017, "Age"].mean()
```

###### 2. Plot the total number of points (`PTS`) per year since 2000.

```python
df_filtered = df[df["Year"] >= 2000]
df_filtered.groupby("Year")["PTS"].sum().sort_index(ascending=True).plot(kind="bar")
```

###### 3. Plot the number of players per year since 2010. What is the year with the most players?

```python
df[df["Year"] > 2010].groupby("Year")["Player"].count().sort_index(ascending=False).plot(kind="bar")
```

```python
df.groupby("Year")["Player"].count()
```

```python
# Gibt das Jahr (z. B. 2024) als einzelnen Wert zurück
id = df.groupby("Year")["Player"].count().idxmax()
print(id)
```
